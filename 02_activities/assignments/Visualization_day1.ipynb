```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(613)

x = np.arange(50)
y = np.random.randint(0, 100, 50)

# Define font styles
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 20}
font2 = {'family': 'monospace', 'color': 'green', 'size': 14}

# Create figure and axis
fig, ax = plt.subplots(figsize=(5, 3))

# Plot
ax.plot(x, y)

# Add labels and title with fonts
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)

plt.tight_layout()
plt.show()
```

```python
np.random.seed(613)

x = np.arange(50)
y = np.random.randint(0,100,50)
# Create figure and axis
fig, ax = plt.subplots(figsize=(5,3))
#set your color and marker sign
ax.scatter(x,y, marker = '*', color = 'indigo')
```

```python
fig, ax = plt.subplots(figsize=(5,3))
#set a style and width to your line
ax.plot(x,y, 
        marker = '*', 
        color = 'indigo',
        linestyle = '--',
        linewidth = 2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
        marker = '*', 
        markersize = 12,
        color = 'indigo',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')
ax.grid(axis = 'y')
```

```python
import matplotlib.pyplot as plt

# create data: an array of values
size_of_groups=[25,8,20,38]

colors = ['skyblue', 'green', 'purple', 'gold']

# Create a pieplot
plt.pie(size_of_groups, colors = colors)
plt.show()
```

```python
import matplotlib.pyplot as plt
 
# create random data
names='groupA', 'groupB', 'groupC', 'groupD',
values=[12,11,3,30]
 
# Create a pieplot
plt.pie(size_of_groups)

# Label distance: gives the space between labels and the center of the pie
plt.pie(values, labels=names, labeldistance=1.15);
plt.show();
```
