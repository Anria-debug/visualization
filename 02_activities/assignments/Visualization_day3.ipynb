```python
np.random.seed(613)
x1 = np.arange(50)
y2 = np.random.randint(0,75,50)
x2 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y2 = np.array([110,180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols = 2,
                               nrows = 1,
                               figsize = (7,3))


ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10,3))
ax1.scatter(x1, y1, color='green')
ax1.set_title('Plot Title')
ax1.set_xlabel('X Axis Label')
ax1.set_ylabel('Y Axis Label')

ax2.bar(x2,y2, color = 'orange')
ax2.set_title('Chart Title')
ax2.set_xlabel('Categories')
ax2.set_ylabel('Values')
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7,4))

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7,4),
                                    )

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize=18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize =18)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7,4),
                                    layout = 'constrained')

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize=18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize =18)
```

```python

x = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1= np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

fig, ax= plt.subplots(figsize=(7,3))
ax.bar(x,y1, color = 'indigo')
ax.plot(x,y2, color = 'red')
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')
ax.errorbar(x, 
            y2,
            yerr = y2_sd,
            fmt = 'none',
            ecolor = 'indigo',
            elinewidth = 4,
            capsize =6,
            capthick = 4)
```

```python
from PIL import Image
import requests 
from io import BytesIO
```

```python
response = requests.get('https://static.wikia.nocookie.net/the-ultimate-good/images/3/3d/Luffy.png/revision/latest?cb=20230417200546')
```

```python
image_file = BytesIO(response.content)
image = Image.open(image_file)
```
